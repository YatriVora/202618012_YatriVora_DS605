## Task 2-Vectorized Array and Linear Algebra

```python
# 2.1 One 2D array and one 3D array (built with np.arange) —
# addition, element-wise multiplication, matrix multiplication
M1 = np.arange(1, 10).reshape(3, 3)          # 2D array, shape (3, 3): [[1,2,3],[4,5,6],[7,8,9]]
M2 = np.arange(1, 19).reshape(2, 3, 3)       # 3D array, shape (2, 3, 3): two stacked 3x3 matrices

print("M1 shape:", M1.shape, "\n", M1)
print("\nM2 shape:", M2.shape, "\n", M2)

addition = M1 + M2                 # M1 broadcasts across M2's first axis -> shape (2, 3, 3)
elementwise_mult = M1 * M2         # same broadcasting -> shape (2, 3, 3)
matmul_result = M1 @ M2            # batched matrix multiplication -> shape (2, 3, 3)

print("\nAddition (M1 + M2):\n", addition)
print("\nElement-wise multiplication (M1 * M2):\n", elementwise_mult)
print("\nMatrix multiplication (M1 @ M2):\n", matmul_result)
```

```python
# 2.2 Transpose and determinant of a square matrix
square = np.array([[4, 7, 2],
                    [3, 6, 1],
                    [5, 2, 8]], dtype=float)

transpose = square.T
determinant = np.linalg.det(square)

print("Square matrix:\n", square)
print("\nTranspose:\n", transpose)
print("\nDeterminant:", determinant)
```

```python
# 2.3 Inverse and verification with np.allclose()
if determinant != 0:
    inverse = np.linalg.inv(square)
    identity_check = square @ inverse
    is_identity = np.allclose(identity_check, np.eye(square.shape[0]))

    print("Inverse:\n", inverse)
    print("\nsquare @ inverse:\n", identity_check)
    print("\nVerified square @ inverse == Identity:", is_identity)
else:
    print("Matrix is singular; inverse does not exist.")
```

## Task 3- Normal Distribution and Histogram

```python
# Task 3 — Normal Distribution and Histogram (using the Generator API)
rng = np.random.default_rng(42)  # seeded Generator for reproducibility

chosen_mean = 50
chosen_std = 10
n_samples = 1000

normal_data = rng.normal(loc=chosen_mean, scale=chosen_std, size=n_samples)

sample_mean = np.mean(normal_data)
sample_std = np.std(normal_data)

print(f"Chosen mean: {chosen_mean}, Chosen std: {chosen_std}, N: {n_samples}")
print(f"Sample mean: {sample_mean:.4f}")
print(f"Sample std : {sample_std:.4f}")
print(f"Difference in mean: {abs(sample_mean - chosen_mean):.4f}")
print(f"Difference in std : {abs(sample_std - chosen_std):.4f}")
```

```python
plt.figure(figsize=(8, 5))
plt.hist(normal_data, bins=30, color="steelblue", edgecolor="black", alpha=0.8)
plt.axvline(sample_mean, color="red", linestyle="--", label=f"Sample mean = {sample_mean:.2f}")
plt.title(f"Histogram of 1000 samples ~ N({chosen_mean}, {chosen_std}^2)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.tight_layout()
plt.show()
```

## Task 5-Filtering and Querying

```python
# Male passengers older than 50
male_over_50 = df[(df["Sex"] == "male") & (df["Age"] > 50)]
print("Male passengers older than 50:", len(male_over_50))
```

```python
# Female first-class passengers: count and survival percentage
female_pclass1 = df[(df["Sex"] == "female") & (df["Pclass"] == 1)]
female_pclass1_survival_pct = female_pclass1["Survived"].mean() * 100
print("Female 1st-class passengers:", len(female_pclass1))
print(f"Survival percentage: {female_pclass1_survival_pct:.2f}%")
```

```python
# Age 20-40, Fare above overall median, and survived
fare_median = df["Fare"].median()
cond5_3 = df[(df["Age"] >= 20) & (df["Age"] <= 40) &
             (df["Fare"] > fare_median) & (df["Survived"] == 1)]
print("Age 20-40, Fare above median, survived:", len(cond5_3))
```

```python
# Travelling alone (SibSp=0, Parch=0), age < 30, and did not survive
cond5_4 = df[(df["SibSp"] == 0) & (df["Parch"] == 0) &
             (df["Age"] < 30) & (df["Survived"] == 0)]
print("Alone, age < 30, did not survive:", len(cond5_4))
```

```python
# Embarked='S', Pclass 2 or 3, and Fare above Southampton median
southampton_median_fare = df.loc[df["Embarked"] == "S", "Fare"].median()
cond5_5 = df[(df["Embarked"] == "S") & (df["Pclass"].isin([2, 3])) &
             (df["Fare"] > southampton_median_fare)]
print("Embarked=S, Pclass 2/3, Fare above Southampton median:", len(cond5_5))
```

## Task 7-Missing Values and Outliers

```python
# Missing value count and percentage for every column
missing_count = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df)) * 100
missing_summary = pd.DataFrame({"Missing_Count": missing_count, "Missing_Percent": missing_pct})
missing_summary = missing_summary[missing_summary["Missing_Count"] > 0].sort_values(
    "Missing_Count", ascending=False)
print("Missing values summary:\n", missing_summary)
```

```python
# Plot missing counts as a bar chart
plt.figure(figsize=(8, 5))
missing_summary["Missing_Count"].plot(kind="bar", color="salmon", edgecolor="black")
plt.title("Missing Value Counts by Column")
plt.ylabel("Count")
plt.xlabel("Column")
plt.tight_layout()
plt.show()
```

```python
# Fill missing Age with mean Age; show before/after
print("Missing Age BEFORE:", df["Age"].isnull().sum())

df_mean_filled = df.copy()
df_mean_filled["Age"] = df_mean_filled["Age"].fillna(df_mean_filled["Age"].mean())

print("Missing Age AFTER (mean fill):", df_mean_filled["Age"].isnull().sum())
```

```python
# Try imputing with mean, median, mode, and a random value drawn from existing Age values
age_mean = df["Age"].mean()
age_median = df["Age"].median()
age_mode = df["Age"].mode()[0]

df_impute = df.copy()
df_impute["Age_mean_imputed"] = df["Age"].fillna(age_mean)
df_impute["Age_median_imputed"] = df["Age"].fillna(age_median)
df_impute["Age_mode_imputed"] = df["Age"].fillna(age_mode)

# Random-value imputation: fill NaNs with random draws from the observed (non-null) Age values
observed_ages = df["Age"].dropna().values
n_missing = df["Age"].isnull().sum()
random_fill_values = rng.choice(observed_ages, size=n_missing, replace=True)

df_impute["Age_random_imputed"] = df["Age"].copy()
missing_mask = df_impute["Age_random_imputed"].isnull()
df_impute.loc[missing_mask, "Age_random_imputed"] = random_fill_values

print(df_impute[["Age", "Age_mean_imputed", "Age_median_imputed",
                  "Age_mode_imputed", "Age_random_imputed"]].head(10))
```

```python
# Fare: Q1, Q3, IQR, 1.5xIQR bounds, and outlier count
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

fare_outliers = df[(df["Fare"] < lower_bound) | (df["Fare"] > upper_bound)]

print(f"Q1: {Q1:.2f}, Q3: {Q3:.2f}, IQR: {IQR:.2f}")
print(f"Lower bound: {lower_bound:.2f}, Upper bound: {upper_bound:.2f}")
print("Number of Fare outliers:", len(fare_outliers))
```

## Task 8-Features and Pivot table

## Task 9- Visualization and observations

```python
# Correlation heatmap for numerical columns
numeric_cols = ["Survived", "Pclass", "Age", "SibSp", "Parch", "Fare", "FamilySize"]
corr = df[numeric_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

Observation: FamilySize is driven almost entirely by SibSp and Parch, as expected by construction
(corr = 0.89 with SibSp, corr = 0.78 with Parch), but FamilySize itself has only a weak direct correlation with survival (corr = 0.02). 

Age is weakly negatively correlated with survival (corr = -0.08) and shows almost no linear relationship with Fare (corr = 0.10) — age alone is not a strong standalone predictor.
Age is moderately negatively correlated with Pclass (corr = -0.37) and with SibSp (corr = -0.31), suggesting younger passengers tended to travel in lower classes and with more siblings/spouses (e.g., families with children). 

Pclass is negatively correlated with survival (corr = -0.34): passengers in lower-numbered (higher) classes were more likely to survive.
